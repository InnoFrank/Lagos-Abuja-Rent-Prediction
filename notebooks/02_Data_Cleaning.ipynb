```python
# Create a copy of the original dataset
df_clean = df.copy()
```

```python
df
```

```python
df_clean["Price"].head(10)
```

```python
df_clean["Price"] = (
    df_clean["Price"]
    .astype(str)
    .str.replace("₦", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.replace("/year", "", regex=False)
    .str.strip()
)
```

```python
df_clean["Price"] = pd.to_numeric(df_clean["Price"], errors="coerce")
```

```python
df_clean["Price"].head(10)
```

```python
df_clean["Price"].dtype
```

```python
df_clean["Price"].isnull().sum()
```

```python
df_clean.to_csv("../data/processed/rent_clean_v1.csv", index=False)
```

```python
df_clean["Bedrooms"] = df_clean["Bedrooms"].str.extract(r"(\d+)")
```

```python
df_clean["Bedrooms"] = pd.to_numeric(df_clean["Bedrooms"], errors="coerce")
```

```python
df_clean["Bedrooms"].head(10)
```

```python
df_clean["Bedrooms"].dtype
```

```python
df_clean["Bathrooms"] = df_clean["Bathrooms"].str.extract(r"(\d+)")
```

```python
df_clean["Bathrooms"] = pd.to_numeric(df_clean["Bathrooms"], errors="coerce")
```

```python
df_clean["Toilets"] = df_clean["Toilets"].str.extract(r"(\d+)")
```

```python
df_clean["Toilets"] = pd.to_numeric(df_clean["Toilets"], errors="coerce")
```

```python
df_clean.dtypes
```

```python
df_clean[["Price", "Bedrooms", "Bathrooms", "Toilets"]].isnull().sum()
```

```python
df_clean = df_clean.dropna(subset=["Price"])
```

```python
df_clean.shape
```

```python
df_clean["Bedrooms"].median()
```

```python
df_clean["Bedrooms"] = df_clean["Bedrooms"].fillna(df_clean["Bedrooms"].median())
```

```python
df_clean["Bathrooms"] = df_clean["Bathrooms"].fillna(df_clean["Bathrooms"].median())
```

```python
df_clean["Toilets"] = df_clean["Toilets"].fillna(df_clean["Toilets"].median())
```

```python
df_clean[["Price", "Bedrooms", "Bathrooms", "Toilets"]].isnull().sum()
```

```python
df_clean.to_csv("../data/processed/rent_clean_v2.csv", index=False)
```
