# Saving, Loading, and Using Your Trained Model

```python
# Import the Required Libraries
import joblib
import pandas as pd
import numpy as np
```

```python
# Load the Saved Model
model = joblib.load(
    "../models/best_rent_prediction_model.pkl"
)
```

```python
# Confirm the Model Loaded
print(model)
```

```python
# Load the Feature Names
feature_names = joblib.load(
    "../models/feature_names.pkl"
)
```

```python
feature_names
```

```python
# Create an Empty Input Row
new_house = pd.DataFrame(
    np.zeros((1, len(feature_names))),
    columns=feature_names
)
```

```python
new_house.head()
```

```python
# Fill in the House Information
new_house["Bedrooms"] = 3
new_house["Bathrooms"] = 3
new_house["Toilets"] = 3

new_house["Serviced"] = 1
new_house["Newly Built"] = 1
new_house["Furnished"] = 1
```

```python
# Deleting those incorrect columns because it's not in the training model, was dropped.
new_house = new_house.drop(
    columns=["Location_Lekki", "Location_Lekki Phase 1"],
    errors="ignore"
)
```

```python
# Set the Location
new_house["Location_Lekki Phase 1 Lekki Lagos"] = 1
```

```python
[col for col in model.feature_names_in_ if "Lekki" in col]
```

```python
# Make a Prediction
prediction = model.predict(new_house)

print(prediction)
```

```python
# Format the Result
print(f"Predicted Rent: ₦{prediction[0]:,.0f}")
```

```python
# Trying Different Houses
new_house["Bedrooms"] = 2
new_house["Bathrooms"] = 2
new_house["Toilets"] = 2
```

```python
prediction = model.predict(new_house)

print(f"Predicted Rent: ₦{prediction[0]:,.0f}")
```

```python
# Create a Reusable Prediction Function
def predict_rent(new_house_data):
    prediction = model.predict(new_house_data)
    return prediction[0]
```

```python
rent = predict_rent(new_house)

print(f"Predicted Rent: ₦{rent:,.0f}")
```
